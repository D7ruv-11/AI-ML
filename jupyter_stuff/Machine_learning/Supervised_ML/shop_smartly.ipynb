# Encoding catagorical variables

```python
from sklearn.preprocessing import OneHotEncoder

# 1. RESET: If 'df' became a sparse matrix, this forces it back to a clean DataFrame
if not isinstance(df, pd.DataFrame):
    # This recovers your data if it was accidentally overwritten by the encoder output
    df = pd.read_csv("shop_smart_ecommerce.csv") 

# 2. Define columns
cols = ["Weekend", "VisitorType", "Month"]

# 3. Initialize Encoder with sparse_output=False
ohe = OneHotEncoder(drop="first", sparse_output=False)

# 4. Transform - we use .astype(str) on thespecific columns only
encoded = ohe.fit_transform(df[cols].astype(str))

# 5. Create the new numeric DataFrame
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(cols), index=df.index)

# 6. Join back and update
df = pd.concat([df.drop(columns=cols), encoded_df], axis=1)

# Verify the result
df.head()
```

```python
x = df.drop("Revenue",axis=1) 
y = df["Revenue"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print("f1 score = ",f1_score(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'max_depth': range(1, 20)}
grid_search = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='f1')
grid_search.fit(x_train, y_train)

print(f"The best depth is: {grid_search.best_params_['max_depth']}")
```

```python
# Train without limit
full_model = DecisionTreeClassifier()
full_model.fit(x_train, y_train)

# Check how deep it actually went
print(f"The tree built itself to a depth of: {full_model.get_depth()} levels")
print(f"Number of leaf nodes: {full_model.get_n_leaves()}")
```

# Hyper-Parameter Tuning 

```python
full_model = DecisionTreeClassifier()
full_model.fit(x_train, y_train)
```

```python
path=full_model.cost_complexity_pruning_path(x_train,y_train)
ccp_alphas = path.ccp_alphas
print(ccp_alphas)
```

```python
f1= []
for i in ccp_alphas:
    model = DecisionTreeClassifier(random_state=42,ccp_alpha=i)
    model.fit(x_train, y_train)
    y_pred =model.predict(x_test)
    f1.append(f1_score(y_test,y_pred))
```

```python
best_alpha = ccp_alphas[f1.index(max(f1))]
print("Best alpha:", best_alpha)
print("Best F1:", max(f1))
```

```python
best_model = DecisionTreeClassifier(random_state=42,ccp_alpha=best_alpha,)
best_model.fit(x_train, y_train)
```

```python
y_prediction = best_model.predict(x_test)
best_f1=f1_score(y_test,y_prediction)
print(best_f1)
```

```python
print("Train F1:", f1_score(y_train,best_model.predict(x_train)))
print("Test F1:", f1_score(y_test, best_model.predict(x_test)))
```
