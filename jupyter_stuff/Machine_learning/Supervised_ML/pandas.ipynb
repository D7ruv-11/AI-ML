```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))
# index
print(s[0])   #index based access
print(s[2])
```

```python
age = pd.Series([23,34,45,12,19], index=["Dhruv","vasu","tanay","shourya","pankhuri"])
print(age)
print(age["Dhruv"])  #label based access
print(age.index)
```

```python
s1 = pd.Series([1,3,5,6,7])
s2 = pd.Series([3,5,6,8,9])
print(s1 + s2)    #vectorized operation
s1[0] = 100   #index value mutuble
print(s1 + s2) 
ch=s1.drop([0])
print(ch)
print(s1)
      
```

```python
#Data frame
#creating Data Frame from  dictionary
info = {
    "name" : ["Dhruv","pankhuri", "vasu"],
    "Age" :[19,20 ,18],
     "CGPA" : [9.5,9.8,10]
}
df = pd.DataFrame(info) 
print(df, type(df))

print(df.index)
print(df.columns)
```

```python
#Data Frame (list in list method)

df = pd.DataFrame([["Dhruv",19],["pankhuri",18],["vasu", 20]],columns = ["name","age"])
print(df)
```

```python
#Data Frame (np.array method)
import numpy as np
np_arr = np.array([[1,2,3],[2,4,6]])
df = pd.DataFrame(np_arr, columns= ["A","B","C"])
print(df)
```
